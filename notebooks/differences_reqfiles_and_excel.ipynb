```python
from learning_driver_preferences.paths import OUTPUT, DATA_DIR
from learning_driver_preferences.plot_style import CUSTOM_COLORS, set_plot_style
from scripts.route_and_request_analysis import * 

import pandas as pd
```

```python
csv_df, csv_info = read_inputfile(OUTPUT / "index_files.csv")
excel_df, excel_info = read_inputfile(DATA_DIR / "ModifiedQueryRows.xlsx", depot = "0521")
```

```python
print(f"rows in csv: {csv_df.shape[0]}")
print(f"rows in excel: {excel_df.shape[0]}")
```

```python
# -------------------------------------------------------
# Compare route IDs
# -------------------------------------------------------
route_ids_only_in_excel = sorted(set(excel_df["route_id"]) - set(csv_df["route_id"]))
route_ids_only_in_csv   = sorted(set(csv_df["route_id"]) - set(excel_df["route_id"]))

# -------------------------------------------------------
# Compare dates
# -------------------------------------------------------
dates_only_in_excel = sorted(set(excel_df["date"]) - set(csv_df["date"]))
dates_only_in_csv   = sorted(set(csv_df["date"]) - set(excel_df["date"]))

# -------------------------------------------------------
# Row‑level comparison (RouteId, Date, Time)
# -------------------------------------------------------
excel_keys = set(zip(excel_df["route_id"], excel_df["date"], excel_df["request_time"]))
csv_keys   = set(zip(csv_df["route_id"], csv_df["date"], csv_df["request_time"]))

rows_in_excel_not_csv = excel_keys - csv_keys
rows_in_csv_not_excel = csv_keys - excel_keys

# -------------------------------------------------------
# Print results
# -------------------------------------------------------
print("\n=== RouteIds only in Excel")
print(route_ids_only_in_excel)

print("\n=== RouteIds only in request files")
print(route_ids_only_in_csv)

print("\n=== Dates only in Excel")
print(dates_only_in_excel)

print("\n=== Dates only in request files ")
print(dates_only_in_csv)

print("\n=== Rows in Excel but NOT in request files - see csv-file for all missing request files")
print(len(list(rows_in_excel_not_csv)))

df_missing_in_request_files = pd.DataFrame(
    list(rows_in_excel_not_csv),
    columns=["route_id", "date", "request_time"]
)

df_missing_in_request_files.to_csv(OUTPUT / "rows_in_excel_not_in_request_files.csv", index=False)

print("\n=== Rows in CSV but NOT in Excel")
print(len(list(rows_in_csv_not_excel)))

```
